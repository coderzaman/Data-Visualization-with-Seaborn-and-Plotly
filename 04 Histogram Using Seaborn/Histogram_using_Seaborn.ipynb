```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a Figure with 2 empty Axes (boxes) side-by-side
fig, ((ax1, ax2), (ax3, ax4), (ax5, ax6)) = plt.subplots(3, 2, figsize=(12, 12))

# Tell Seaborn to put the scatterplot in the FIRST box (ax1)
sns.scatterplot(data=student, x="study_hours", y="test_score", ax=ax1)

# Tell Seaborn to put the lineplot in the SECOND box (ax2)
sns.lineplot(data=student, x="week", y="study_hours", ax=ax2, errorbar=None)

sns.lineplot(data=student, x="week", y="test_score", ax=ax3, errorbar=None)

sns.histplot(data=student, x="test_score", ax=ax4)

sns.lineplot(data=student, x="week", y="test_score", ax=ax5, errorbar=None)

sns.histplot(data=student, x="test_score", ax=ax6)
plt.tight_layout()
plt.show()
```
