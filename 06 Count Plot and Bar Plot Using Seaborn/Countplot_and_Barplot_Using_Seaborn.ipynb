# **Countplot**

A **Count Plot** is one of the most frequently used charts in data analysis.

Simply put, a Count Plot is exactly like a histogram, but instead of grouping *numbers* into bins, it counts the frequency of *categories* (words/text).

Think of it as an automatic tally counter. You give it a column of categorical data (like "Gender", "City", or "Grade"), and Seaborn automatically counts how many times each category appears and draws a bar for it.

### How it works visually
To really understand what the code is doing behind the scenes, play with this interactive Count Plot. Add different students to the dataset and watch how the chart automatically tallies them up and adjusts the bars:


### How to do it in Python
Writing a count plot in Seaborn is incredibly simple because you only need to provide **one axis** (either X or Y). Seaborn handles the counting for the other axis automatically.

Here is the exact code to create one:

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. A simple dataset
data = {
    'Student_Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah'],
    'Class_Level': ['Freshman', 'Sophomore', 'Freshman', 'Junior', 'Freshman', 'Senior', 'Sophomore', 'Freshman'],
    'Gender': ['Female', 'Male', 'Male', 'Male', 'Female', 'Male', 'Female', 'Female']
}
df = pd.DataFrame(data)

# 2. Set the theme
sns.set_theme(style="darkgrid")

# 3. Create the Count Plot
# Notice we only provide x="Class_Level". Seaborn calculates the Y-axis (Count) for us!
sns.countplot(data=df, x="Class_Level")

plt.title("Number of Students per Class Level")
plt.show()
```

### Pro-Tip: Adding `hue` for extra detail
Just like in `scatterplot` or `lineplot`, you can use the `hue` parameter in a count plot to split the bars into sub-categories.

If you change the plotting line to this:
```python
sns.countplot(data=df, x="Class_Level", hue="Gender")
```
Seaborn will now draw *two* bars for each class level (one for Male, one for Female), allowing you to easily compare the gender breakdown within each class!

Do you want to see how to automatically sort these bars from tallest to shortest, or how to write the exact number on top of each bar?

# **Bar Plot**

Here is the easiest way to remember the difference:
* **`sns.countplot()`** answers: *"How **many** students are in each class?"* (It only needs an X-axis, it calculates the Y-axis by counting).
* **`sns.barplot()`** answers: *"What is the **average** test score for each class?"* (It needs both an X-axis and a Y-axis).

### How Seaborn's Bar Plot Works Under the Hood
When you use a standard bar chart in Matplotlib, it usually just plots exactly the numbers you give it.

Seaborn’s `barplot` is much smarter. It is a **statistical chart**. If you give it a dataset where there are 20 Male students and 20 Female students, and you ask it to plot their Test Scores, Seaborn will automatically:
1. Group all the Male scores together.
2. Calculate the **Mean (Average)** of those scores.
3. Draw a bar up to that average number.
4. Draw a little black line on top of the bar called an **Error Bar**.

### The Mystery of the Error Bar
Beginners are always confused by the little black line on top of a Seaborn bar plot.

Because Seaborn is showing you an *average* of many students, it uses that line to tell you how much the data varies. A short error bar means everyone got roughly the same score. A long error bar means the scores were all over the place (some very high, some very low), so the "average" might not be super reliable.

### How to do it in Python

Here is the code to create a Bar Plot. Notice that unlike the Count Plot, we must provide both `x` and `y`!

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. A dataset with multiple test scores per subject
data = {
    'Subject': ['Math', 'Math', 'Math', 'Science', 'Science', 'Science'],
    'Test_Score': [80, 85, 90, 50, 75, 100],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}
df = pd.DataFrame(data)

# 2. Set the theme
sns.set_theme(style="whitegrid")

# 3. Create the Bar Plot
# Seaborn automatically calculates the average Test_Score for each Subject
sns.barplot(data=df, x="Subject", y="Test_Score")

plt.title("Average Test Score by Subject")
plt.show()
```

**Pro-Tip:** Just like before, you can add `hue="Gender"` to this code, and Seaborn will automatically calculate the average Male score and average Female score, placing those bars side-by-side!

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. A dataset with multiple test scores per subject
data = {
    'Subject': ['Math', 'Math', 'Math', 'Science', 'Science', 'Science'],
    'Test_Score': [80, 85, 90, 50, 75, 100],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}
df = pd.DataFrame(data)

# 2. Set the theme
sns.set_theme(style="whitegrid")

# 3. Create the Bar Plot
# Seaborn automatically calculates the average Test_Score for each Subject
sns.barplot(data=df, x="Subject", y="Test_Score", hue='Gender', errorbar=None)

plt.title("Average Test Score by Subject")
plt.show()
````
